```python
import os
import cv2
import numpy as np
import pandas as pd

from tqdm import tqdm

import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
```

```python
model_path = r"C:\Users\varun\Downloads\New folder (2)\models\mobilenet_v3_small.tflite"
```

```python
from mediapipe.tasks import python
from mediapipe.tasks.python import vision
```

```python
dataset_path = r"C:\Users\varun\Downloads\New folder (2)\dataset\dataset_o"
```

```python
base_options = python.BaseOptions(
    model_asset_path=model_path
)

options = vision.ImageEmbedderOptions(
    base_options=base_options,
    l2_normalize=True,
    quantize=False
)

embedder = vision.ImageEmbedder.create_from_options(options)
```

```python
import os
print(os.listdir(dataset_path))
```

```python
rows = []

classes = sorted(os.listdir(dataset_path))

for label in classes:

    folder = os.path.join(dataset_path, label)

    if not os.path.isdir(folder):
        continue

    print("Processing:", label)

    for file in tqdm(os.listdir(folder)):

        path = os.path.join(folder, file)

        image = cv2.imread(path)

        if image is None:
            continue

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

        mp_image = mp.Image(
            image_format=mp.ImageFormat.SRGB,
            data=image
        )

        result = embedder.embed(mp_image)

        if len(result.embeddings) == 0:
            continue

        emb = result.embeddings[0]

        if hasattr(emb, "float_embedding"):
            features = list(emb.float_embedding)

        elif hasattr(emb, "embedding"):
            features = list(emb.embedding)

        elif hasattr(emb, "numpy_view"):
            features = emb.numpy_view().tolist()

        else:
            print("Unknown embedding format")
            print(dir(emb))
            exit()

        features.append(label)

        rows.append(features)

print("Saving CSV...")

columns = [f"feature_{i}" for i in range(len(rows[0])-1)]
columns.append("label")

df = pd.DataFrame(rows, columns=columns)

df.to_csv("landmarker.csv", index=False)

print(df.head())

print("CSV Saved Successfully!")
print("Shape:", df.shape)
```
